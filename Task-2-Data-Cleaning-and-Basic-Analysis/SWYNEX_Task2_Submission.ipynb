#### 2.5.2 Classify Tumor and Non-Tumor Samples

```python
import csv
import pandas as pd
import gzip

# Extract sample IDs and sample titles from GEO file
with gzip.open(file_path, "rt") as file:
    for line in file:
        if line.startswith("!Sample_geo_accession"):
            sample_ids = next(csv.reader([line], delimiter="\t"))[1:]

        elif line.startswith("!Sample_title"):
            sample_titles = next(csv.reader([line], delimiter="\t"))[1:]

# Create metadata table
metadata = pd.DataFrame({
    "Sample_ID": sample_ids,
    "Sample_Title": sample_titles
})

# Classify tissue types
metadata["Tissue_Group"] = metadata["Sample_Title"].apply(
    lambda x: "Tumor" if x.startswith("Lung Tumor")
    else "Non-Tumor" if x.startswith("Normal Lung")
    else "Unknown"
)

# Display first 10 samples
metadata.head(10)
```

```python
print("Total samples:", len(metadata))

print("\nTissue group counts:")
print(metadata["Tissue_Group"].value_counts())

print("\nUnknown classifications:",
      (metadata["Tissue_Group"] == "Unknown").sum())

print("\nSample IDs match expression matrix:",
      metadata["Sample_ID"].tolist() == df.columns.tolist())
```

#### Interpretation – Sample Metadata Extraction and Classification
Sample metadata was extracted from the GEO Series Matrix file (GSE10072). A metadata table was created by matching GEO sample accession IDs with their corresponding sample titles.

Based on the sample titles, the 107 samples were classified into two groups:

Tumor: 58 samples

Non-tumor: 49 samples

No unknown classifications were found. The sample IDs in the metadata table also matched the expression matrix columns in the same order.

Therefore, the sample classification was successfully verified, allowing further comparative analysis between tumor and non-tumor lung tissues.

#### 2.5.3 Separate Expression Matrices

```python
# Identify sample IDs belonging to each tissue group
tumor_samples = metadata.loc[
    metadata["Tissue_Group"] == "Tumor", "Sample_ID"
].tolist()

non_tumor_samples = metadata.loc[
    metadata["Tissue_Group"] == "Non-Tumor", "Sample_ID"
].tolist()

# Separate expression data
tumor_df = df[tumor_samples]
non_tumor_df = df[non_tumor_samples]

# Check dimensions
print("Tumor expression matrix:", tumor_df.shape)
print("Non-tumor expression matrix:", non_tumor_df.shape)
```

#### 2.5.4 Group-wise Mean Comparison

```python
# Calculate mean expression for each group
tumor_mean = tumor_df.mean(axis=1)
non_tumor_mean = non_tumor_df.mean(axis=1)

# Create a comparison table
group_comparison = pd.DataFrame({
    "Tumor_Mean": tumor_mean,
    "Non_Tumor_Mean": non_tumor_mean
})

# Display first 10 probes
group_comparison.head(10)
```

```python
# Calculate difference in mean expression
group_comparison["Mean_Difference"] = (
    group_comparison["Tumor_Mean"] -
    group_comparison["Non_Tumor_Mean"]
)

# Display first 10 probes
group_comparison.head(10)
```

#### Biological Interpretation
The mean expression values were calculated separately for tumor (58 samples) and non-tumor lung tissue (49 samples).

The difference in mean expression was calculated by subtracting the non-tumor mean from the tumor mean for each probe.

Positive mean differences indicate higher average expression signals in tumor samples, whereas negative values indicate higher average signals in non-tumor samples. Values close to zero indicate relatively similar average signals between the two groups.

For example, probe 1007_s_at showed a positive mean difference of approximately 0.463, while probe 1294_at showed a negative mean difference of approximately -0.521.

These observations provide an initial comparison of expression patterns between tumor and non-tumor tissues. Further statistical analysis is required to determine whether these differences are statistically significant.
